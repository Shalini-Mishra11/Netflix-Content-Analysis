```python
type_counts = df['type'].value_counts()

type_counts
```

```python
plt.figure(figsize=(6,4))

type_counts.plot(kind='bar')

plt.title('Movies vs TV Shows')
plt.xlabel('Type')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(6,6))

type_counts.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.ylabel('')

plt.title('Movies vs TV Shows')

plt.show()
```

```python
country_df = df[df["country"] != "Unknown"]
```

```python
country_df["country"].value_counts().head(10)
```

```python
country_counts = country_df["country"].value_counts().head(10)
```

```python
plt.figure(figsize=(10,5))

country_counts.plot(kind='bar')

plt.title('Top 10 Countries on Netflix')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)

plt.show()
```

```python
rating_counts = df["rating"].value_counts()

print(rating_counts.head(10))
```

```python
plt.figure(figsize=(10,5))

rating_counts.head(10).plot(kind="bar")

plt.title("Top Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()

print(year_counts.tail())
```

```python
plt.figure(figsize=(12,5))

year_counts.plot(kind="line")

plt.title("Netflix Content by Release Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
director_df = df[df["director"] != "Unknown"]

top_directors = director_df["director"].value_counts().head(10)

print(top_directors)
```

```python
plt.figure(figsize=(10,5))

top_directors.plot(kind="bar")

plt.title("Top 10 Directors")
plt.xlabel("Director")
plt.ylabel("Number of Titles")

plt.xticks(rotation=75)

plt.show()
```

```python
movies = df[df["type"] == "Movie"]
```

```python
movies = df[df["type"] == "Movie"].copy()
```

```python
movies["duration_min"] = (
    movies["duration"]
    .str.replace(" min", "", regex=False)
    .astype(int)
)
```

```python
plt.figure(figsize=(10,5))

plt.hist(movies["duration_min"], bins=30)

plt.title("Distribution of Movie Duration")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

```python
movies["duration_min"].mean()
```

```python
movies["duration_min"].min()
```

```python
movies["duration_min"].max()
```

```python
movies.sort_values(
    by="duration_min",
    ascending=False
)[["title", "duration_min"]].head(10)
```

```python
genres = (
    df['listed_in']
    .str.split(', ')
    .explode()
)
```

```python
genre_counts = genres.value_counts().head(10)

genre_counts
```

```python
plt.figure(figsize=(10,5))

genre_counts.plot(kind='bar')

plt.title("Top 10 Genres on Netflix")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```
